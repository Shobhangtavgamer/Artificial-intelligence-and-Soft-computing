# **1. Fuzzy Set Operations (Union, Intersection, Complement)**

```python
# Define fuzzy sets with elements and their membership degrees
A = {"x1": 0.2, "x2": 0.7, "x3": 0.4, "x4": 1.0}
B = {"x1": 0.5, "x2": 0.3, "x3": 0.9, "x4": 0.0}

# Perform Fuzzy Union (Max), Intersection (Min), and Complement (1 - x)
fuzzy_union = {k: max(A[k], B[k]) for k in A}
fuzzy_intersection = {k: min(A[k], B[k]) for k in A}
fuzzy_complement_A = {k: round(1.0 - A[k], 1) for k in A}

print("Fuzzy Set A:      ", A)
print("Fuzzy Set B:      ", B)
print("Union (A U B):    ", fuzzy_union)
print("Intersect (A n B):", fuzzy_intersection)
print("Complement (~A):  ", fuzzy_complement_A)
```

# **2. Single-Layer Perceptron (AND Gate)**

```python
import numpy as np

# Inputs for AND gate and corresponding targets
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])

# Initialize random weights, bias, and learning rate
weights = np.array([0.0, 0.0])
bias = 0.0
lr = 0.1

# Training loop over 10 epochs
for epoch in range(10):
    for inputs, target in zip(X, y):
        # Step function activation: 1 if linear combination >= 0 else 0
        linear_combination = np.dot(inputs, weights) + bias
        prediction = 1 if linear_combination >= 0 else 0

        # Update weights and bias based on error
        error = target - prediction
        weights += lr * error * inputs
        bias += lr * error

# Test final trained output
print("Trained Weights:", weights, "Bias:", bias)
print("Predictions for AND gate:")
for inputs in X:
    pred = 1 if (np.dot(inputs, weights) + bias) >= 0 else 0
    print(f"Input: {inputs} -> Predicted Output: {pred}")
```

# **3. Triangular Membership Function**

```python
def triangular_membership(x, a, b, c):
    """Calculates fuzzy membership value for a triangular distribution."""
    if x <= a or x >= c:
        return 0.0
    elif a < x <= b:
        return (x - a) / (b - a)
    elif b < x < c:
        return (c - x) / (c - b)
    return 0.0


# Test with parameters: a=2, b=5, c=8
a, b, c = 2, 5, 8
test_inputs = [1, 3.5, 5, 6.5, 9]

print(f"Triangular Function (a={a}, b={b}, c={c}):")
for x in test_inputs:
    degree = triangular_membership(x, a, b, c)
    print(f"  Input x = {x:<4} -> Membership Degree = {degree:.2f}")
```

# **4. Trapezoidal Membership Function**

```python
def trapezoidal_membership(x, a, b, c, d):
    """Calculates fuzzy membership value for a trapezoidal distribution."""
    if x <= a or x >= d:
        return 0.0
    elif b <= x <= c:
        return 1.0
    elif a < x < b:
        return (x - a) / (b - a)
    elif c < x < d:
        return (d - x) / (d - c)
    return 0.0


# Test with parameters: a=1, b=3, c=6, d=8
a, b, c, d = 1, 3, 6, 8
test_inputs = [0, 2, 4.5, 7, 9]

print(f"Trapezoidal Function (a={a}, b={b}, c={c}, d={d}):")
for x in test_inputs:
    degree = trapezoidal_membership(x, a, b, c, d)
    print(f"  Input x = {x:<4} -> Membership Degree = {degree:.2f}")
```

# **5. McCulloch-Pitts Neuron (Logic Gates)**

```python
def mc_culloch_pitts(inputs, weights, threshold):
    # Calculate the linear combination of inputs and weights
    net_input = sum(x * w for x, w in zip(inputs, weights))
    # Threshold firing rule (Output is 1 if net_input >= threshold, else 0)
    return 1 if net_input >= threshold else 0


# 1. Simulate AND Gate (Weights =, Threshold = 2)
# 2. Simulate OR Gate  (Weights =, Threshold = 1)
inputs_list = [[0, 0], [0, 1], [1, 0], [1, 1]]

print("AND Gate Simulation:")
for x in inputs_list:
    output = mc_culloch_pitts(x, weights=[1, 1], threshold=2)
    print(f"  Input: {x} -> Output: {output}")

print("\nOR Gate Simulation:")
for x in inputs_list:
    output = mc_culloch_pitts(x, weights=[1, 1], threshold=1)
    print(f"  Input: {x} -> Output: {output}")
```

# **6. Heuristic Search Algorithm (Greedy Best-First Search).**

```python
# Graph represented as an adjacency list: (Neighbor, Actual Edge Cost)
graph = {
    "A": [("B", 1), ("C", 4)],
    "B": [("D", 5)],
    "C": [("D", 1)],
    "D": [],
}

# Straight-Line Distance Heuristic values h(n) to target node 'D'
heuristic = {"A": 6, "B": 4, "C": 1, "D": 0}


def greedy_best_first(start, goal):
    """Finds a path to the goal by always expanding the node with lowest h(n)."""
    visited = set()
    queue = [[start]]  # Queue stores paths

    while queue:
        # Sort queue so the path ending with the lowest heuristic value comes first
        queue.sort(key=lambda path: heuristic[path[-1]])
        path = queue.pop(0)
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)
            for neighbor, cost in graph.get(node, []):
                new_path = list(path)
                new_path.append(neighbor)
                queue.append(new_path)
    return None


# Execute search from 'A' to 'D'
solution = greedy_best_first("A", "D")
print(f"Greedy Best-First Search Path: {solution}")
```

# 7.  A* Search Algorithm

```python
# Graph layout: { Node: [(Neighbor, Edge_Cost), ...] }
graph = {
    "A": [("B", 1), ("C", 4)],
    "B": [("D", 5)],
    "C": [("D", 1)],
    "D": [],
}

# Heuristic values h(n) estimating distance to goal 'D'
heuristic = {"A": 6, "B": 4, "C": 1, "D": 0}


def a_star_search(start, goal):
    """Finds the optimal path from start to goal minimizing f(n) = g(n) + h(n)."""
    # Items in queue format: (total_estimated_cost_f, current_actual_cost_g, [path_nodes])
    queue = [(heuristic[start], 0, [start])]
    visited = set()

    while queue:
        # Sort queue to ensure we always expand the path with the lowest f(n)
        queue.sort(key=lambda x: x[0])
        f, g, path = queue.pop(0)
        node = path[-1]

        if node == goal:
            return path, g

        if node not in visited:
            visited.add(node)
            for neighbor, cost in graph.get(node, []):
                new_g = g + cost
                new_f = new_g + heuristic[neighbor]
                queue.append((new_f, new_g, path + [neighbor]))
    return None, float("inf")


# Execute A* search from 'A' to 'D'
optimal_path, total_cost = a_star_search("A", "D")
print(f"Optimal A* Path: {optimal_path}")
print(f"Total Path Cost: {total_cost}")
```

# **8. Associative Memory Network (Hopfield Network).**

```python
import numpy as np

# Define binary patterns as flat arrays using bipolar values (-1 and 1)
pattern_A = np.array([1, 1, -1, -1])

# 1. Training Phase: Compute Weight Matrix using Hebbian Learning Rule
# W = outer_product(A, A) and set the diagonal elements to 0 (no self-connection)
W = np.outer(pattern_A, pattern_A)
np.fill_diagonal(W, 0)


def recall(test_pattern, weights, steps=5):
    """Asynchronously or synchronously updates state to recall the stored memory."""
    state = np.array(test_pattern)
    for _ in range(steps):
        # Calculate network activation energy and apply hard-threshold sign function
        raw_activation = np.dot(weights, state)
        state = np.where(raw_activation >= 0, 1, -1)
    return state


# Test data: Introduce noise into pattern_A (flip the last element from -1 to 1)
noisy_input = np.array([1, 1, -1, 1])
recovered_output = recall(noisy_input, W)

print("Stored Pattern (Memory):", pattern_A)
print("Noisy Input Provided:   ", noisy_input)
print("Recovered Memory Output:", recovered_output)
```

# **9. Fuzzy Bounded Sum and Difference**

```python
# Bounded sum: min(1, A + B) | Bounded difference: max(0, A + B - 1)
A = {"x1": 0.6, "x2": 0.4, "x3": 0.9}
B = {"x1": 0.5, "x2": 0.2, "x3": 0.3}

bounded_sum = {k: min(1.0, A[k] + B[k]) for k in A}
bounded_diff = {k: max(0.0, round(A[k] + B[k] - 1.0, 1)) for k in A}

print("Fuzzy Set A:        ", A)
print("Fuzzy Set B:        ", B)
print("Bounded Sum:        ", bounded_sum)
print("Bounded Difference: ", bounded_diff)
```

# **10. Genetic Algorithm (Crossover and Mutation Loop)**

```python
import random

# Configuration
chromosome_length = 8
mutation_rate = 0.1

# Generate two random binary chromosomes (Parents)
parent1 = [random.randint(0, 1) for _ in range(chromosome_length)]
parent2 = [random.randint(0, 1) for _ in range(chromosome_length)]


def crossover_and_mutate(p1, p2):
    # Single-point crossover: swap bits after a random index split point
    split = random.randint(1, chromosome_length - 1)
    child = p1[:split] + p2[split:]

    # Bit-flip mutation: invert bit if random probability hits mutation rate
    mutated_child = [
        bit if random.random() > mutation_rate else 1 - bit for bit in child
    ]
    return child, mutated_child, split


child, mutated, split_point = crossover_and_mutate(parent1, parent2)
print("Parent 1:       ", parent1)
print("Parent 2:       ", parent2)
print(f"Crossover Child (Split at index {split_point}):", child)
print("Mutated Child:  ", mutated)
```

# **11. Hebbian Learning Rule (Unsupervised Weight Update)**

```python
import numpy as np

# Bipolar Input pattern vector X and resulting Bipolar Output y
X = np.array([1, -1, 1])
y = 1
learning_rate = 1.0
initial_weights = np.array([0.0, 0.0, 0.0])


def hebbian_update(weights, inputs, output, lr):
    """Adjusts weights based on synchronous activation."""
    # Delta W = learning_rate * input * output
    delta_w = lr * inputs * output
    new_weights = weights + delta_w
    return new_weights, delta_w


final_w, change = hebbian_update(initial_weights, X, y, learning_rate)
print("Initial Weights: ", initial_weights)
print("Weight Delta:    ", change)
print("Updated Weights: ", final_w)
```

# **12. Breadth-First Search (BFS Graph Traversal)**

```python
# Graph dictionary representation
graph = {"A": ["B", "C"], "B": ["D", "E"], "C": ["F"], "D": [], "E": [], "F": []}


def bfs_traverse(start_node):
    visited = []
    queue = [start_node]

    while queue:
        node = queue.pop(0)  # Dequeue first element (FIFO)
        if node not in visited:
            visited.append(node)
            # Append all unvisited neighbors to the back of the queue
            queue.extend([n for n in graph[node] if n not in visited])
    return visited


print("BFS Traversal Order starting from A:", bfs_traverse("A"))
```
